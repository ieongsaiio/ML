# Data Summary

The data used in this project is the dataset released by the Kaggle "Ames Housing Price" competition. The original source is the residential assessment records collected by housing tax auditors in Ames, Iowa, USA from 2006 to 2010. After being sorted out, it is released for academic and teaching use.
The dataset contains 1 row of ID, and 79 feature fields related to the target variable, where the target variable is sale price. There is total 81 columns (including the target column) and 1460 observations. The fields can be divided into the following types:
numerical continuous type, such as LotArea (land area) and GrLivArea (ground living area); ordinal discrete type, which represents the quality levels, such as OverallQual, KitchenQual; and nominal classification, which describes the things such as materials and neighbourhoods.
It is noticed that there are missing values in fields. Most of them are referred to "None" or 0 as missing markers, not the real recording error.


# Objective of the Analysis

The core task of this analysis is to build an interpretable and generalizable regression model to predict the final selling price of homes in Ames and gain insight into the main factors that affect the price. The implementation is as follows: Exploratory and Data Analysis (EDA), dataset cleaning, features engineering and selection, model training and evaluation

# Model Comparison

I compared regularized linear models and OLS. Ridge selected α≈0 (reducing to OLS) and Elastic Net selected l1_ratio≈1 (reducing to Lasso). Lasso and OLS achieve similar test R² around 0.92. For prediction, Lasso is slightly better; for inference, I use OLS with HC3‑robust standard errors.

# Key Findings

Neighborhood premiums are large: StoneBr and Crawfor add about $27k and $21k, respectively; size and space matter: GrLivArea and TotalBsmtSF are positive, and a small positive GrLivArea² indicates a convex size–price curve; LuxuryScore and LuxuryScore² show accelerating premiums for high‑end homes; quality × size interaction (Luxury_HighQual × GrLivArea) is also positive; amenities like CentralAir, Brick veneer, and Basement exposure increase price; older houses and more bedrooms (holding size fixed) reduce price.

# Limitations and Next Steps

Residuals are non‑normal and heteroscedastic; HC3 fixes SEs but not the shape, and errors are larger for very cheap and very expensive homes. Inference is post‑selection, so p‑values can be biased. Next, using other models rather than just linear regression with regularization may be a better choice.
